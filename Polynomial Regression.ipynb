# Polynomial Regression

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_absolute_error,r2_score
from sklearn.linear_model import LinearRegression
```

```python
url="https://raw.githubusercontent.com/AdiPersonalWorks/Random/master/student_scores%20-%20student_scores.csv"
df=pd.read_csv(url)
df
```

```python
plt.scatter(df["Hours"],df["Scores"])
plt.title("Hours Vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```

```python
x=df[["Hours"]]
y=df['Scores']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=30)
```

```python
poly=PolynomialFeatures(degree=2)
x_train_poly=poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
model = LinearRegression()
model.fit(x_train_poly,y_train)
y_pred=model.predict(x_test_poly)
```

```python
print("R2 Score:",r2_score(y_test,y_pred))
print("Mean Absolute Error:",mean_absolute_error(y_test,y_pred))
```

```python
# visualize polynomial regression
x_plot=np.linspace(x.min(),x.max(),100).reshape(-1,1)
x_plot_poly=poly.transform(x_plot)
y_plot=model.predict(x_plot_poly)
```

```python
plt.scatter(x,y,color='blue',label="Actual data")
#x_sorted=np.sort(x,axis=0)
#y_sorted_pred= model.predict(poly.transorm(x_sorted))
plt.plot(x_plot,y_plot,color='red',label='Polynomial regression curve')
plt.title("Polynomial regression stufent score")
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```

```python
hours=[[7.5]]
hours_poly=poly.transform(hours)
predicted_score = model.predict(hours_poly)
print(f"Predicted score for studying {hours[0][0]} hours = {predicted_score[0]:.2f}")
```
